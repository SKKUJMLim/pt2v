```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

def add_gaussian_noise(image, sigma):
    """
    image: PIL Image
    sigma: Gaussian noise std (0 ~ 255 scale)
    """
    img = np.array(image).astype(np.float32)
    noise = np.random.normal(0, sigma, img.shape)
    noisy_img = img + noise
    noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)
    return Image.fromarray(noisy_img)
```

```python
# 이미지 로드
img = Image.open("../basketball.jpg").convert("RGB")

# 노이즈 강도 설정
sigmas = [5, 100, 200, 300]

noisy_images = [add_gaussian_noise(img, s) for s in sigmas]

# 시각화
plt.figure(figsize=(12, 3))
for i, (s, ni) in enumerate(zip(sigmas, noisy_images)):
    plt.subplot(1, len(sigmas), i+1)
    plt.imshow(ni)
    plt.title(f"σ = {s}")
    plt.axis("off")
plt.show()
```

```python
def diffusion_forward(image, num_steps=4, max_sigma=80):
    """
    image: PIL Image
    num_steps: number of diffusion steps
    max_sigma: maximum noise level
    """
    img = image
    sigmas = np.linspace(0, max_sigma, num_steps)

    results = []
    for sigma in sigmas:
        noisy = add_gaussian_noise(img, sigma)
        results.append((sigma, noisy))

    return results
```

```python
steps = diffusion_forward(img, num_steps=4, max_sigma=400)

plt.figure(figsize=(12, 3))
for i, (s, im) in enumerate(steps):
    plt.subplot(1, len(steps), i+1)
    plt.imshow(im)
    #plt.title(f"Step {i}\nσ={int(s)}")
    plt.axis("off")
plt.show()
```
